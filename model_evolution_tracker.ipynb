# 📈 ASAG Model Evolution Tracker

This notebook visualizes the journey of our Automated Student Answer Grading (ASAG) models, from the initial classical baseline to the final state-of-the-art Mixture of Experts (MoE) architecture.

## 1. The Data: Recording Our Milestones
Based on the project history, we track four major milestones in model development and their corresponding accuracies.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set visual style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Define the project evolution milestones
milestones = [
    {
        "Phase": "1. Baseline (Classical)",
        "Model": "TF-IDF + SVM",
        "Accuracy": 72.0,
        "Description": "Basic keyword matching. Fast but ignores semantics."
    },
    {
        "Phase": "1.5 Baseline (Embeddings)",
        "Model": "Frozen SBERT + SVM",
        "Accuracy": 77.0,
        "Description": "Captured basic meaning but limited by frozen vocabulary."
    },
    {
        "Phase": "2. Neural Era (Fine-Tuning)",
        "Model": "Fine-tuned MiniLM Transformer",
        "Accuracy": 85.8,
        "Description": "Retrained to 'understand' grading nuance specifically."
    },
    {
        "Phase": "3. The Final Architecture",
        "Model": "Mixture of Experts (MoE)",
        "Accuracy": 93.2,
        "Description": "Dedicated specialists (FR/AR) solving language interference."
    }
]

df = pd.DataFrame(milestones)
display(df)
```

## 2. Visualizing the Trajectory
Let's plot this growth to clearly see the leap in performance our architectural choices provided.

```python
# Create the plot
fig, ax = plt.subplots(figsize=(12, 7))

# Plot points and connecting line
sns.lineplot(data=df, x='Phase', y='Accuracy', marker='o', markersize=12, linewidth=3, color='#2c3e50', ax=ax)

# Fill the area underneath for visual impact
ax.fill_between(df['Phase'], df['Accuracy'], alpha=0.1, color='#3498db')

# Add data labels
for i, row in df.iterrows():
    ax.text(i, row['Accuracy'] + 0.8, f"{row['Accuracy']}%", 
            ha='center', va='bottom', weight='bold', fontsize=12, color='#e74c3c')
    
    # Add Model names below points
    ax.text(i, row['Accuracy'] - 1.5, row['Model'], 
            ha='center', va='top', fontsize=10, style='italic', color='#7f8c8d')

# Styling
ax.set_title('🚀 Evolution of ASAG Model Accuracy', fontsize=18, weight='bold', pad=20)
ax.set_xlabel('Project Phase', fontsize=14, weight='bold', labelpad=15)
ax.set_ylabel('Accuracy (%)', fontsize=14, weight='bold', labelpad=15)
ax.set_ylim(60, 100)

# Modify x-axis ticks for better readability
plt.xticks(rotation=45, ha='right', fontsize=11)

plt.tight_layout()
plt.show()
```

## 🎯 Conclusion
The journey from a standard TF-IDF matcher to a bilingual Mixture of Experts yielded an absolute accuracy improvement of **+21.2%**. 

The most significant jumps occurred when we:
1. **Unfroze the model (85.8%):** Allowing the model to learn the *task* of grading.
2. **Split the architecture (93.2%):** Removing language interference by giving French and Arabic their own dedicated Expert brains.
